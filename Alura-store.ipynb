# **Análisis de facturación**

## **Ingreso total por tienda**

```python
tienda_1['Tienda'] = 'Tienda 1'
tienda_2['Tienda'] = 'Tienda 2'
tienda_3['Tienda'] = 'Tienda 3'
tienda_4['Tienda'] = 'Tienda 4'

tiendas = pd.concat([tienda_1, tienda_2, tienda_3, tienda_4], ignore_index=True)
```

```python
tiendas.sample(5)
```

```python
ingresos_totales = (
    tiendas.groupby('Tienda')['Precio']
    .sum()
    .map('{:,.2f}'.format)
    .sort_values(ascending=True)
)
ingresos_totales
```

## **Ventas por categoría**

```python
tiendas['Categoría del Producto'].unique()
```

```python
ventas_categoria = (
    tiendas.groupby('Tienda')['Categoría del Producto']
    .value_counts()
    .reset_index(name='Ventas')
)
ventas_categoria
```

```python
top_categorias = (
    ventas_categoria
    .sort_values(['Tienda', 'Categoría del Producto'], ascending=[True, False])
    .groupby('Tienda')
    .head(3)
)

top_categorias
```

```python
bottom_categorias = (
    ventas_categoria
    .sort_values(['Tienda', 'Categoría del Producto'], ascending=[True, True])
    .groupby('Tienda')
    .head(3)
)
bottom_categorias
```

## **Valoración media por tienda**

```python
#Tienda mejor valorada
valoración_media = (
    tiendas.groupby('Tienda')['Calificación']
    .mean()
    .reset_index(name='Calificación')
    .sort_values('Calificación', ascending=False)
    .set_index('Tienda')
)
valoración_media
```

## **Productos más y menos vendidos**

```python
# Lista de todos los productos que se venden en las 4 tiendas
productos_totales = tiendas['Producto'].unique()
productos_totales
```

```python
ventas_producto = (
    tiendas.groupby(['Tienda', 'Producto'])
    .size()
    .reset_index(name = 'Ventas')
)
ventas_producto.head(20)
```

```python
mas_vendidos = (
    ventas_producto
    .sort_values(['Tienda', 'Ventas'], ascending=[True, False])
    .groupby('Tienda')
    .head(10)
)

mas_vendidos
```

```python
menos_vendidos = (
    ventas_producto
    .sort_values(['Tienda', 'Ventas'], ascending=[True, True])
    .groupby('Tienda')
    .head(10)
)

menos_vendidos
```

```python
resumen = pd.merge(
    mas_vendidos,
    menos_vendidos,
    on = 'Tienda',
    suffixes = ('Más Vendidos', 'Menos Vendidos')
)
resumen.sample(10)
```

```python
#Top 5 productos más vendidos por tienda
top5 = (
    ventas_producto
    .sort_values(['Tienda', 'Ventas'], ascending=[True, False])
    .groupby('Tienda')
    .head(5)
)

for tienda in top5['Tienda'].unique():

    df_tienda = top5[top5['Tienda'] == tienda]

    plt.figure()
    plt.bar(df_tienda['Producto'], df_tienda['Ventas'])
    plt.title(f'Top 5 productos más vendidos - {tienda}')
    plt.xticks(rotation=45)
    plt.ylabel('Número de ventas')

    plt.savefig(f'top5_productos_{tienda}.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
#Top 5 productos menos vendidos por tienda
bottom5 = (
    ventas_producto
    .sort_values(['Tienda', 'Ventas'], ascending=[True, True])
    .groupby('Tienda')
    .head(5)
)

for tienda in bottom5['Tienda'].unique():

    df_tienda = bottom5[bottom5['Tienda'] == tienda]

    plt.figure()
    plt.bar(df_tienda['Producto'], df_tienda['Ventas'])
    plt.title(f'Top 5 productos menos vendidos - {tienda}')
    plt.xticks(rotation=45)
    plt.ylabel('Número de ventas')

    plt.savefig(f'bottom5_productos_{tienda}.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
for tienda in ventas_producto['Tienda'].unique():
    df_tienda = ventas_producto[ventas_producto['Tienda'] == tienda]
    
    masvendido = df_tienda.loc[df_tienda['Ventas'].idxmax()]
    menosvendido = df_tienda.loc[df_tienda['Ventas'].idxmin()]

    print('--------------------------------------------')
    print('Tienda:', tienda)
    print('El producto más vendido es:', masvendido['Producto'], 'con', masvendido['Ventas'], 'unidades vendidas')
    print('El producto menos vendido es:', menosvendido['Producto'], 'con', menosvendido['Ventas'], 'unidades vendidas')
```

## **Valor del envío promedio por tienda**

```python
costo_envio_promedio = (
    tiendas.groupby('Tienda')['Costo de envío']
    .mean()
    .reset_index()
    .sort_values('Costo de envío', ascending=False)
)
costo_envio_promedio
```

```python
fig, ax = plt.subplots(figsize=(12, 8))

tienda = ingresos_totales.axes[0].tolist()
ingreso = ingresos_totales.values.tolist()
bar_colors = ['orange', 'forestgreen', 'orchid', 'purple']

ax.bar(tienda, ingreso, color=bar_colors)

ax.set_xlabel('Tienda')
ax.set_ylabel('Ingresos Totales')
ax.set_title('Ingresos Totales por Tienda')
ax.invert_xaxis()

plt.savefig('ingresos_tienda.png', dpi=300, bbox_inches='tight')
plt.show()
```

## **Productos más y menos vendidos**

```python
pivot_mas = top5.pivot_table(
    index='Producto',
    columns='Tienda',
    values='Ventas',
    fill_value=0
)

pivot_mas.plot(kind='bar', stacked=True, figsize=(10, 6))

plt.title('Top productos más vendidos (comparación por tienda)')
plt.ylabel('Ventas')
plt.xticks(rotation=90)

plt.savefig('top_productos.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
pivot_menos = bottom5.pivot_table(
    index='Producto',
    columns='Tienda',
    values='Ventas',
    fill_value=0
)

pivot_menos.plot(kind='bar', stacked=True, figsize=(10, 6))

plt.title('Top productos menos vendidos (comparación por tienda)')
plt.ylabel('Ventas')
plt.xticks(rotation=90)

plt.savefig('bottom_productos.png', dpi=300, bbox_inches='tight')
plt.show()
```

## **Métodos de Pago por cada tienda**

```python
pagos_tienda = (
    tiendas.groupby(['Tienda', 'Método de pago'])
    .size()
    .reset_index(name = 'Pagos')
)

for tienda in pagos_tienda['Tienda'].unique():

    df_tienda = pagos_tienda[pagos_tienda['Tienda'] == tienda]

    plt.figure(figsize=(6,6))

    plt.pie(
        df_tienda['Pagos'],
        labels=df_tienda['Método de pago'],
        autopct='%1.1f%%',
        startangle=90,
        textprops={'fontsize': 10},
        pctdistance=0.9
    )

    plt.title(f'Distribución de métodos de pago {tienda}')

    plt.savefig(f'metodos_pago_{tienda}.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
ubicacion_tiendas = tiendas.groupby(['Tienda', 'lat', 'lon']).size().reset_index(name='Cantidad de transacciones')

plt.figure(figsize=(10,6))

sns.scatterplot(data=ubicacion_tiendas, 
                x='lon', 
                y='lat', 
                hue='Tienda',
                style='Tienda', 
                palette='Set2', 
                size='Cantidad de transacciones',
                sizes=(100,300),
                alpha=0.7)
plt.title(f'Ubicación geográfica de las 4 tiendas')
plt.xlabel('Longitud')
plt.ylabel('Latitud')

plt.savefig(f'ubicacion_tiendas.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.kdeplot(
    data=ubicacion_tiendas,
    x='lon',
    y='lat',
    fill=True
)

plt.title('Mapa de calor: concentración de ventas')
plt.xlabel('Longitud')
plt.ylabel('Latitud')

plt.savefig('concentracion_ventas.png', dpi=300, bbox_inches='tight')
plt.show()
```

# **Informe final: Evaluación de Tiendas**

**INTRODUCCIÓN**

El propósito de este análisis es ayudar al señor Juan a tomar una decisión informadda sobre qué tienda debería ser descartada, pues no está alcanzando los niveles de productividad esperados en términos de rentabilidad, demanda y costos operativos.

Para ello, se evaluaron sus cuatro tiendas, utilizando indicadores clave como:
- Ingresos totales
- Categorías de productos más y menos vendidas
- Calificaciones promedio de satisfacción de los clientes
- Productos más y menos vendidos
- Coste de envío promedio
- Métodos de pago
- Desempeño geográfico

**DESARROLLO DEL ANÁLISIS**

**- Ingresos totales por tienda**
Se calcularon los ingresos totales generados por cada tienda, con el fin de medir cual tiene mayor capacidad de ventas.

**Interpretación:**
- La tienda con mayores ingresos fue la **Tienda 1**, lo cual indica una alta demanda y mayor volumen de ventas.
- Por otro lado, la **Tienda 4** presento menores ingresos, lo cual podría reflejar menor flujo de clientes o menor desempeño comercial.



**- Categorías de productos más y menos vendidas**

Se analizaron las categorías con mayor y menor numero de ventas en cada tienda.

**Hallazgos**
- En las cuatro tiendas las categorías de productos más vendidas son **Muebles, Libros y Juguetes**.
- En las cuatro tiendas las categorías de productos menos vendidas son **Artículos para el Hogar, Deportes y Diversión y Electrodomésticos**
- Dado que en las 4 tiendas, tanto las categorías más y menos vendidas son idénticas, esta variable parecería no tener relevancia para la toma de decisión, pues en las 4 se venden las mismas categorías en proporciones similares.



**- Calificación promedio de clientes por tienda**

Se evaluó el nivel de satisfacción del cliente mediante las calificaciones promedio.

**Resultados**
- La tienda con mejor valoración fue la **Tienda 3**, con una calificación promedio de **4.04**, mientras que la peor valorada fue la **Tienda 1**, lo que podría implicar problemas de servicio o de logística. 
- En el caso de la calificación baja en la **Tienda 1** podría deberse a que esta tienda también es la que mayor **Costos de envío** presenta. Dichos costos son financiados por los clientes, por lo que esa podría ser la razón de la calificación obtenida.



**- Productos más y menos vendidos**

Se identificaron los productos con mayor y menor volumen de ventas.

**Interpretación**
- Los productos más vendidos de la tienda con más rendimiento (Tienda 1) fueron: **Armario, Microondas, TV LED UHD 4K, Secadora de Ropa y Bloques de Construcción**.
- Los productos menos vendidos de la Tienda 1 fueron: **Auriculares con Micrófono, Celular ABXY, Olla de Presión y Pandereta**
- Los productos más vendidos de la tienda con menor desempeño en ventas (Tienda 4) fueron: **Cama Box, Cubertería, Cama King, Dashboards y Carrito de Control Remoto**
- Los productos menos vendidos de la Tienda 4 fueron: **Guitarra Eléctrica, Armario, Guitarra Acústica, Ciencia de Datos y Lavadora de Ropa** 
- Esto indica las tendencias de consumo en ambas tiendas y la baja demanda de dichos artículos en cada tienda respectivamente, lo cual permitiría saber en qué productos invertir más para aumentar su inventario y en cuales no.
- Este análisis permite entender qué tan dinámico es el catálogo de la tienda.



**- Valor del envío promedio**

El coste del envío promedio es un factor clave porque afecta directamente el precio final para el cliente y la competitividad de la tienda.

**Resultados**
- La tienda con menor coste promedio de envío fue la **Tienda 4** con una inversión por parte de los clientes de 23459.45 pesos, lo cual representa una ventaja logística.
- Una tienda con costos altos podría desincentivar compras, especialmente en productos de bajo precio.
- Por el contario, la **Tienda 1** es la que tiene mayor costo de envío. Aunque el envío es más costoso, esto no ha afectado negativamente las ventas, lo que sugiere que los clientes perciben un alto valor en los productos ofrecidos por esta tienda.
- Estos resultados muestran una asociación entre mayores ventas y mayores costos de envío, pero no necesariamente implican una relación causal directa.



**- Métodos de pago más utilizados en cada tienda**

Se analizaron los métodos de pago más utilizados en cada tienda para ver si alguna estaría absorbiendo el financiamiento de algunos productos por el uso de cupones de descuento.

**Hallazgos**
- La tienda en la que los clientes han usado más cupones es la **Tienda 2**, sin embargo esto no ha impactado de gran manera en los ingresos puesto que sólo el 6% de las ventas son pagadas con este método y aún así es la segunda tienda con más ingresos.



**- Análisis del desempeño geográfico**

Se hizo un mapa de calor con los datos geográficos de la tienda, donde se representa la concentración de ventas en función de la longitud (eje x) y la latitud (eje y).

**Resultados**
- En la imagen se ven tres zonas principales acorde a la cantidad de ventas realizadas.
- Se observa una región dominante alrededor de (-75, 5), lo que sugiere que la mayor parte de las transacciones se agrupan en esa zona.
- También se identifican regiones secundarias con menor intensidad, lo que podría representar mercados adicionales.
- Sin embargo, debido a que el dataset presenta coordenadas repetidas para distintas tiendas, el análisis geográfico debe interpretarse con cautela, ya que la distribución espacial puede no corresponder a ubicaciones reales.

**CONCLUSIÓN**

Con base en el análisis realizado sobre el desempeño de las cuatro tiendas, se evaluaron distintos factores clave con el objetivo de identificar cuál presenta el rendimiento más bajo y, por lo tanto, representa la mejor opción para que el Sr. Juan considere venderla o desprenderse de ella.

A lo largo del estudio se compararon los ingresos totales, las categorías de productos más y menos vendidas, los productos con mayor y menor demanda, las calificaciones promedio otorgadas por los clientes, así como el costo de envío promedio en cada tienda. Estos indicadores permiten obtener una visión integral del comportamiento comercial de cada sucursal.

Los resultados obtenidos muestran que la Tienda 4 presenta un desempeño consistentemente inferior en comparación con las demás. En particular, registra los menores ingresos totales, un volumen reducido de ventas y una menor concentración de productos destacados en las categorías más demandadas. Esto sugiere que su contribución al rendimiento general del negocio es limitada.

Adicionalmente, aunque la Tienda 4 presenta el costo de envío promedio más bajo, este factor no se traduce en un mejor desempeño comercial. Esto indica que su bajo rendimiento no está determinado por los costos logísticos, sino posiblemente por una menor demanda, menor competitividad en su oferta de productos o una experiencia de compra menos favorable para los clientes.

En contraste, otras tiendas logran mantener mayores niveles de ventas e ingresos incluso con costos de envío más elevados, lo cual refleja una operación más sólida y una mayor aceptación por parte de los consumidores.

Por lo tanto, considerando que el objetivo del Sr. Juan es desprenderse de la tienda con menor desempeño, se concluye que la Tienda 4 es la opción más adecuada para ser vendida, ya que representa la unidad menos rentable y con menor impacto comercial dentro del conjunto analizado.

Esta recomendación se encuentra respaldada por los datos obtenidos y las visualizaciones generadas, las cuales evidencian de forma clara las fortalezas y debilidades comparativas de cada tienda.
